```python
cover_median = df["cover"].sort_values().at[7]
print("Median cover type:", cover_median)
```

Cover Column Median

The median in the cover column for books is pb or paperback which tells us that the middle of that column we will get a paperback book.

Calculating Mean Weight

```python
df["weight"].mean()
```

```python
all_cover = df.groupby("cover")["weight"].mean()
print(all_cover.round(2))
```

Based on the calculations it shows us that hardback books weigh about 796 grams but paperback books weight about 628 grams. This tells us that hardback books weigh more than the paperback type of books.

Calculating the Difference

```python
hardback = df[df["cover"] == "hb"]["weight"].mean()
paperback = df[df["cover"] == "pb"]["weight"].mean()

difference = hardback - paperback

print("Hardback mean:", round(hardback, 2), "grams")
print("Paperback mean:", round(paperback, 2), "grams")
print("Difference:", round(difference, 2), "grams")
```

Recommendation

Based on the weight of the books I would choose the paperback books. They would be easier to put away because of how much they weight. Compared to the hardback books the paperback books weigh about 168 grams less which helps the people at the library.
