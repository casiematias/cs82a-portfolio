Module 4 Lab 3 Trucks Price

```python
import pandas as pd
df = pd.read_csv("truck.csv")
df.head()
```

```python
r = df["X"].corr(df["Y"])
print("Pearson correlation coefficient:", r)
print("Pearson correlation coefficient:", round(r, 3))
```

Pearson's r

After the calculations for Pearson's r we can see that r is very close to 1. When r is close to +1 that means there is a strong positive linear relationship. The X and Y column for the trucks price shows that when one increases the other price will increase as well.

Scatter Plot

```python
plot = df.plot.scatter(x="X", y="Y")
plot.set_title("Truck List Price vs. Best Sale Price")
plot.set_xlabel("List Price by $1,000")
plot.set_ylabel("Best Sale Price by $1000")
```

Prediction

When the sale price is 25,500 and we make an estimate on how much the sale price will be but, we have to be careful. The scatter plot we have only goes up to 22,400 which is not the best but Pearson's r is equal to 0.999 which is about 99.99% accurate. Based on r and how positive the relationship is when the list price is 25,500 the sale price will be about or above 21,000.
