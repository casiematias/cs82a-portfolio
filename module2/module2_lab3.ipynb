Module 2 Lab 3 Temperature Changes

Categories:
Cold - temperatures below 15°C
Warm - temperatures from 15°C to 25°C (inclusive)
Hot - temperatures above 25°C

List of Temperatures
temperatures = [10, 16, 23, 28, 14, 22, 30]

```python
temperatures = [10, 16, 23, 28, 14, 22, 30]
```

```python
for temp in temperatures:
    if temp < 15:
        weather = "Cold"
    elif temp <= 25:
        weather = "Warm"
    else:
        weather = "Hot"

    print(f"{temp}°C: {weather}")
```

This for loop checks the list of temperatures and are categorized based on the
cutoffs. The if, elif, and else in the loop are what used to make sure the numbers
are cutoff at the right point based on the conditions.
