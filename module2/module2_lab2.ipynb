While Loop Multiplication

```python
def while_numbers(a, b):
    final = 0
    count = 0

    while count < b:
        final = final + a
        count = count + 1

    return final

print("7 x 6 =", while_numbers(7, 6))
print("10 x 8 =", while_numbers(10, 8))
print("4 x 9 =", while_numbers(4, 9))
```

The while loop terminates when the count which is dependent of the amount of b is not less than the amount of b. Which is simply saying when the count is equal to b the while loop will terminate.

For Loop Multiplication

```python
def for_numbers(x, y):
    total = 0

    for i in range(y):
        total = total + x

    return total

print("7 x 6 =", for_numbers(7, 6))
print("10 x 8 =", for_numbers(10, 8))
print("4 x 9 =", for_numbers(4, 9))
```

The for loop multiplication will repeat the addition for the amount of y which will get us our result.

Final Test

```python
print("While loop: 7 x 6 =", while_numbers(7, 6))
print("For loop: 7 x 6 =", for_numbers(7, 6))
```

Both loops give us the same result even when using different ways to get there without multiplication. The while loop multiplication is dependent on the b when it comes to the count and will not stop while the count is less than the amount of b. The for loop is dependent on the range of y which is the stopping point but will continue the addition in the range of y.
