```python
temps = [61, 65, 72, 58, 70, 74, 68]
count = 0
for t in temps:
    if t > 65:
        count = count + 1
print(count)
```

```python
def to_celsius(f):
    return (f - 32) * 5/9

print(round(to_celsius(temps[0]), 1))
```
