```python
def urban_group(x):
    if x < 50:
        return "small"
    elif x < 60:
        return "medium"
    elif x < 70:
        return "large"
    else:
        return "extra-large"

df["urban_group"] = df["UrbanPop"].apply(urban_group)
df[["UrbanPop", "urban_group"]].head(10)
```

```python
df.groupby("urban_group")[["Murder", "Assault"]].mean()
```
