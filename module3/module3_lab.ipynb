Cleaning Log

1. Beginning

In order to know what the start point was for the data I called shape and head(10). The shape showed us the 300 rows and 6 rows in the table. For the head(10) call it showed us the preview of 10 rows and what those 10 rows and 6 columns contained.

2. Missing Prices

For the price column in the data there were some missing numbers or float numbers missing. In order to fix that I used the median of the price column because it is not affected by the high or low prices. The 12 prices were filled in with the median of 37.53. So, in the end the missing values were now going to be filled with the median of the price column which gives us a cleaner data.

3. Duplicate Rows

The dataset had 300 rows and so I called duplicated.sums which showed us there were 8 duplicates so that was a problem. In order to fix it I dropped the duplicates which brought the total of rows from 300 to 292 with 6 columns. Now the data is not duplicated and it is much cleaner while still having all the important data still in the dataset.

4. ZIP codes

The zip column had one row where we saw that it was a 4 number ZIP not a regular 5 number ZIP. In order to fix the number we had to make it a 5 number str instead of a 4 so we used strfill(5). This made sure that that if any more ZIP codes had 4 numbers it was now filled with a 0 to make it an accurate ZIP code.

5. Dates

The date column was not shown as the same with some showing the year first and some starting with the month. In order to fix that I used pd.datetime() with a format=mixed which showed us the type as datetime64[us] so that they were all the same type now.

6. Negative Quantities

There were 3 rows with negative number in the quantity column and a decision had to be made. I chose to KEEP the negative numbers since there was an order then that was the amount of product the business was going to lose. The price was a positive number so I read it as we lost a certain amout of product but gained the money from the order.

7. Final Dataset

Now the final dataset has 292 rows and 6 columns with no duplicate rows. The missing prices were now filled with the median which is not affected by the whole of the high and low prices. In the ZIP code column all the ZIP codes are now 5 numbers and the date column all of the dates are now that same type. The negative rows are kept because of my decision with my reasoning.
