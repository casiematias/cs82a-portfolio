Missing Lengths

There are some missing lengths but I would not make a big deal out of it. Since the value is missing I would just set the number to the median just to make sure that there is a value in the length column. The median is never truly affected by the highest and lowest length so it is set as the fill for the lengths of 0.

```python
length_median = df["length"].median()
print(length_median)
```

```python
df["length"] = df["length"].fillna(length_median)
print(df["length"].isna().sum())
```
